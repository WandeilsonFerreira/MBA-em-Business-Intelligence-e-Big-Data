```python
# Nome do bucket no MinIO
bucket_name = "dados-enriquecidos-dw"
# Diretório local com as imagens
local_directory = "planilhas/"

# Função para copiar arquivos de imagem para o MinIO
def upload_images_to_minio(local_directory, bucket_name):
    # Verificar se o bucket existe e criar caso não exista
    if not minio_client.bucket_exists(bucket_name):
        minio_client.make_bucket(bucket_name)
        print(f"Bucket '{bucket_name}' criado com sucesso.")
    else:
        print(f"Bucket '{bucket_name}' já existe.")

    # Iterar sobre os arquivos do diretório local
    for file_name in os.listdir(local_directory):
        # Verificar se o arquivo é uma imagem
        if file_name.endswith(('.xls', '.xlsx', '.csv', '.txt', '.dat')):
            file_path = os.path.join(local_directory, file_name)
            try:
                # Enviar arquivo para o MinIO
                minio_client.fput_object(
                    bucket_name,  # Nome do bucket
                    file_name,    # Nome do objeto no bucket
                    file_path     # Caminho do arquivo local
                )
                print(f"Planilha '{file_name}' copiada com sucesso para o bucket '{bucket_name}'.")
            except S3Error as e:
                print(f"Erro ao copiar '{file_name}': {e}")
```

```python
# Executa a função para copiar as imagens
upload_images_to_minio(local_directory, bucket_name)
```
